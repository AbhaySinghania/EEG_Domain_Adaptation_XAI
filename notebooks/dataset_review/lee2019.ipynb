```python
import mne
from moabb.datasets import Lee2019_MI

mne.set_config('MNE_DATA', '../../Data/moabb_cache')

dataset = Lee2019_MI()
data = dataset.get_data(subjects=[1])
```

```python
print(type(data))
print(data.keys())
```

```python
# Get subject 1 data
subject_data = data[1]
print(type(subject_data))
print(subject_data.keys())
```

```python
# Get session 1 data
session_data = subject_data['1']
print(type(session_data))
print(session_data.keys())
```

```python
# Get the actual data
run_data = session_data['1train']
print(type(run_data))
print(run_data)
```

```python
import mne

raw = run_data
events, event_id = mne.events_from_annotations(raw)
print(event_id)
print(f"Total events found: {len(events)}")
print(events[:10])
```

```python
epochs = mne.Epochs(
    raw,
    events,
    event_id={'left_hand': 1, 'right_hand': 2},
    tmin=0,
    tmax=3,
    baseline=None,
    preload=True
)
print(epochs)
```

```python
data2 = dataset.get_data(subjects=[2])
raw2 = data2[2]['1']['1train']
events2, event_id2 = mne.events_from_annotations(raw2)
epochs2 = mne.Epochs(
    raw2,
    events2,
    event_id={'left_hand': 1, 'right_hand': 2},
    tmin=0,
    tmax=3,
    baseline=None,
    preload=True
)

trial1 = epochs['left_hand'][0].get_data()[0, 0, :]
trial2 = epochs2['left_hand'][0].get_data()[0, 0, :]
times = epochs['left_hand'][0].times

import matplotlib.pyplot as plt
plt.figure(figsize=(12, 5))
plt.plot(times, trial1 * 1e6, label='Subject 1', alpha=0.7)
plt.plot(times, trial2 * 1e6, label='Subject 2', alpha=0.7)
plt.xlabel('Time (seconds)')
plt.ylabel('Amplitude (microvolts)')
plt.title('Lee2019_MI - Same Motor Imagery Task - Subject 1 vs Subject 2')
plt.legend()
plt.grid(True)
plt.show()
```
